## Langchain Basics

LangChain is an open-source framework that helps developers build applications powered by large language models (LLMs) like GPT, Claude, LLaMA, etc.

Instead of just asking a model “What’s 2+2?”, LangChain helps you build complex apps like:

- Chatbots 💬
- Agents that can use tools (SQL, APIs, calculators, search engines) 🔧
- Document search (RAG – Retrieval Augmented Generation) 📄
- Workflow automation (chains of steps) ⚡

### os.getenv and os.environ

- os.getenv: It reads the OPENAI_API_KEY from the system (if it exists).

- os.environ: Then re-sets it into os.environ (so that it’s available to the Python process and libraries).

Think of it like this analogy:

- System environment variable = Permanent storage (set in OS).

- os.environ = Temporary copy available to Python right now.

- os.getenv = Reads the value.

- Assignment to os.environ = Writes/overrides the value in the current process.

#### format_instructions 
is auto-generated guidance from the parser that ensures the LLM responds in the correct JSON structure your JsonOutputParser expects.

```python
from langchain_core.output_parsers import JsonOutputParser
from langchain_core.prompts import ChatPromptTemplate, PromptTemplate
from langchain_openai import ChatOpenAI
from pydantic import BaseModel, Field

# Initialize model
model = ChatOpenAI(temperature=0.7)

# Define Pydantic schema
class Joke(BaseModel):
    setup: str
    punchline: str

# Query
joke_query = "Tell me a joke."

# Set up parser
parser = JsonOutputParser(pydantic_object=Joke)

# ChatPromptTemplate with system + user messages
prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful assistant that tells jokes in JSON format."),
    ("user", "{query}\n{format_instructions}")
])

# Build chain
chain = prompt | model | parser

# Invoke with dictionary input
response = chain.invoke({
    "query": joke_query,
    "format_instructions": parser.get_format_instructions()
})

print(response)
```

```python
from langchain.output_parsers import YamlOutputParser
from langchain_core.prompts import PromptTemplate
from langchain_openai import ChatOpenAI
from pydantic import BaseModel, Field


# Define your desired data structure.
class Joke(BaseModel):
    setup: str = Field(description="question to set up a joke")
    punchline: str = Field(description="answer to resolve the joke")


model = ChatOpenAI(temperature=0.5)

# And a query intented to prompt a language model to populate the data structure.
joke_query = "Tell me a joke."

# Set up a parser + inject instructions into the prompt template.
parser = YamlOutputParser(pydantic_object=Joke)

prompt = PromptTemplate(
    template="Answer the user query.\n{format_instructions}\n{query}\n",
    input_variables=["query"],
    partial_variables={"format_instructions": parser.get_format_instructions()},
)

chain = prompt | model | parser

chain.invoke({"query": joke_query})
```
